## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
## 3. Quick look at the data (2-3 real numbers)
'''
Filter applied per the lane guide: `impressions_90d > 0`, `content_age_days >= 90`, deduplicated by `content_id`.'''
```

```python
import pandas as pd

df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
print('raw rows:', len(df))

# apply the lane guide's filter and dedup rule
df = df.drop_duplicates(subset='content_id')
df = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)]
print('filtered rows:', len(df))

# 1) how common is decline in this data?
pct_down = (df['trend_direction'] == 'down').mean() * 100
n_down = (df['trend_direction'] == 'down').sum()
print(f"\n1) {pct_down:.1f}% of rows ({n_down:,} of {len(df):,}) are labeled trend_direction == 'down'.")

# 2) is decline hitting pages that actually matter (have demand)?
down = df[df['trend_direction'] == 'down']
notdown = df[df['trend_direction'] != 'down']
print(f"\n2) Median impressions_90d — declining pages: {down['impressions_90d'].median():.0f} "
      f"vs. non-declining pages: {notdown['impressions_90d'].median():.0f}.")

# 3) how many candidates would a 'declining_with_demand' reason code surface?
cand = df[(df['trend_direction'] == 'down') & (df['impressions_90d'] >= 100)]
print(f"\n3) {len(cand):,} rows ({len(cand)/len(df)*100:.1f}% of filtered rows) meet the "
      f"'declining_with_demand' reason code (trend down AND impressions_90d >= 100) — "
      f"more candidates than any reviewer could get through by hand.")
```
